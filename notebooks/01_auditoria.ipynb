```python
import pandas as pd
from pathlib import Path
pd.set_option("display.max_rows", 50)

RUTA_BTS = Path("../data/raw/bts")
archivos = sorted(RUTA_BTS.glob("*.csv"))
for a in archivos:
    print(a.name)


# Solo la cabecera del primer mes, para ver los nombres exactos de las columnas
cabecera = pd.read_csv(archivos[0], nrows=0)
print(len(cabecera.columns), "columnas")
print(list(cabecera.columns))
```

```python
#seleccionamos las columnas que queremos para nuestros analisis 
columnas = ["Year", "Month", "DayofMonth", "DayOfWeek", "FlightDate", "Reporting_Airline","Tail_Number", "Flight_Number_Reporting_Airline", "Origin", "OriginCityName","OriginState", "Dest", "DestCityName", "DestState", "CRSDepTime", "DepTime", "DepDelay","DepDel15", "DepTimeBlk", "TaxiOut", "TaxiIn", "CRSArrTime", "ArrTime", "ArrDelay","ArrDel15", "Cancelled", "CancellationCode", "Diverted", "AirTime", "Distance", "CarrierDelay", "WeatherDelay", "NASDelay", "SecurityDelay", "LateAircraftDelay"]
# leemos los csv 
meses_csv = []                     
for archivo in archivos:
    df_mes = pd.read_csv(archivo, usecols=columnas)   
    meses_csv.append(df_mes)                        
    print(archivo.name, df_mes.shape)          

df = pd.concat(meses_csv, ignore_index=True)           
```

Carga de datos. Se cargaron las 35 columnas seleccionadas de los tres CSV mensuales del BTS: enero (539.747 filas), febrero (504.884) y marzo (600.872). Tras concatenarlos, el DataFrame tiene 1.645.503 filas × 35 columnas y ocupa 1,3 GB en memoria, por las 11 columnas de texto (object). Se optimizará con dtype category en la limpieza.

## NULOS EN EL DF

```python
#veremos el porcentaje de nulos por columna 
pct_nulos = df.isna().mean() * 100
pct_nulos = pct_nulos.round(2).sort_values(ascending=False)
print(pct_nulos)
```

## Qué dicen tus nulos 

- CancellationCode (98,14 %): solo se rellena si el vuelo se canceló, así que su nulo es la ausencia de cancelación. Por tanto, aproximadamente 100 - 98,14 = 1,86 % de vuelos cancelados.
- Las 5 columnas de causa (80,72 %): las cinco tienen exactamente el mismo porcentaje porque se rellenan a la vez y solo cuando el retraso de llegada es >15 min. El 19,28 % restante (317.266 vuelos) son los que tienen causa asignada.
- Grupo ~1,8-2,1 % (DepTime, DepDelay, ArrTime, TaxiOut, AirTime...): coincide con el orden de magnitud de las cancelaciones. Los vuelos cancelados nunca despegaron, así que no tienen horas reales.
- Las diferencias entre ellos son pistas: ArrDelay/AirTime (2,09 %) superan a DepTime (1,80 %), lo que sugiere otro grupo además de los cancelados. Mi hipótesis son los desviados (Diverted), que despegan pero no aterrizan en destino.
- Tail_Number (0,25 %): matrícula vacía, sin relación evidente con lo anterior.

## ¿CUantos vuelos hay cancelados o desviados? 
para ello primero tenemos que comprobar los valores que hay 0/1 (1 = vuelo cancelado / desviado

```python
print(df["Cancelled"].unique())
print(df["Diverted"].unique())
#observamos que los valores son los correctos ahira 
##Cuantos vuelos hay cancelados? y desviados? 
mask = (df["Cancelled"] == 1) & (df["Diverted"] == 1)
print(df.loc[mask, "ArrDelay"].isna().mean())
```

```python
# 1) Cancelados: cuántos y qué porcentaje
print(df["Cancelled"].sum(), df["Cancelled"].mean() * 100)

# 2) Desviados: lo mismo
print(df["Diverted"].sum(), df["Diverted"].mean() * 100)
```

Cancelaciones y desvíos. Hay 30.640 vuelos cancelados (1,86 %) y 3.817 desviados (0,23 %). El 1,86 % coincide con el complemento del nulo de CancellationCode (100 - 98,14), luego ese campo solo se informa en vuelos cancelados.

```python
# 3) Vuelos normales (ni cancelados ni desviados)
mask = (df["Cancelled"] == 0) & (df["Diverted"] == 0)
print(mask.sum())                               # cuántas filas tiene la selección
print(df.loc[mask, "ArrDelay"].isna().mean())    # proporción de nulos en ArrDelay
print(df.loc[mask, "DepDelay"].isna().mean())    # y en DepDelay
```

Verificación. Hay 1.611.046 vuelos ni cancelados ni desviados (1.645.503 - 30.640 - 3.817). En ese subconjunto ArrDelay y DepDelay tienen 0 % de nulos. Por tanto, los nulos de estas dos variables proceden exclusivamente de vuelos cancelados o desviados, no de fallos de registro, y no se imputarán.

### diagnóstico de los nulos

```python
# duplicados exactos (todas las columnas)
print(df.duplicated().sum())

# duplicados por identificador natural del vuelo
clave = ["FlightDate", "Reporting_Airline", "Flight_Number_Reporting_Airline", "Origin", "Dest", "CRSDepTime"]
print(df.duplicated(subset=clave).sum())
```

No hay filas duplicadas: df.duplicated() devuelve 0 sobre las 35 columnas, y también 0 usando como identificador natural del vuelo la combinación de fecha, aerolínea, número de vuelo, origen, destino y hora programada de salida (CRSDepTime). Por tanto, no es necesario eliminar registros y los conteos y medias posteriores no están inflados por repeticiones.

### tipos de datos

```python
cols_texto = df.select_dtypes(include="object").columns
print(df[cols_texto].nunique().sort_values())
print(df["FlightDate"].head())
```

- CancellationCode (4), Reporting_Airline (14), DepTimeBlk (19), OriginState/DestState (52): pocos valores distintos, repetidos millones de veces. Candidatas claras a category.
- OriginCityName/DestCityName (327) y Origin/Dest (333): también compensan a category (333 valores en 1,6 millones de filas es una repetición enorme).
- Tail_Number (5.787): son matrículas de aeronave. Aún repite mucho (5.787 valores en 1,6 M de filas), así que category también ahorra memoria, aunque es la menos evidente.
- FlightDate (90): no es una categoría, es una fecha. Enero (31) + febrero (28) + marzo (31) = 90 días, lo que cuadra exactamente con el periodo. Está en formato YYYY-MM-DD, así que se convierte sin ambigüedad con pd.to_datetime.

De las 35 columnas, 11 son de tipo object (texto) y explican casi todo el uso de memoria (1,3 GB). Nueve tienen pocos valores distintos respecto a las 1.645.503 filas (CancellationCode 4, Reporting_Airline 14, DepTimeBlk 19, OriginState y DestState 52, OriginCityName y DestCityName 327, Origin y Dest 333), por lo que se convertirán a category. Tail_Number (5.787 matrículas) también se beneficia de ello. FlightDate (90 valores únicos, coherentes con los 90 días de enero-marzo, formato YYYY-MM-DD) se convertirá a datetime.

### Ahorro de memoria con category

```python
antes = df["Origin"].memory_usage(deep=True)
despues = df["Origin"].astype("category").memory_usage(deep=True)
print(antes / 1e6, despues / 1e6, (1 - despues / antes) * 100)
```

Como prueba, se midió la columna Origin (333 valores distintos en 1.645.503 filas): como object ocupa 98,7 MB y como category 3,3 MB, un ahorro del 96,6 %. Al aplicar la conversión a las columnas de texto de baja cardinalidad se espera una reducción muy notable de los 1,3 GB iniciales. La conversión real se realizará en el notebook de limpieza.
